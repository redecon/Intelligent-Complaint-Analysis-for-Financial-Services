```python
import re
import pandas as pd
import os

# Define target products
TARGET_PRODUCTS = {"Credit card", "Personal loan", "Savings account", "Money transfers"}

# Input and output paths (relative to your notebook location)
RAW_PATH = "../data/raw/complaints.csv"
OUT_PATH = "../data/processed/filtered_complaints.csv"

# Text cleaning function
def clean_text(s: str) -> str:
    if not isinstance(s, str):
        return ""
    t = s.lower()
    t = re.sub(r"[^a-z0-9\s\.,;:\-\?!']", " ", t)
    t = re.sub(r"\s+", " ", t).strip()
    return t

# Preprocessing function
def preprocess(raw_path, out_path):
    df = pd.read_csv(raw_path, low_memory=False)
    df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

    # Standardize product names
    df["product_std"] = df["product"].str.strip().str.title()

    # Flag narratives
    df["narrative_exists"] = df["consumer_complaint_narrative"].notna() & (
        df["consumer_complaint_narrative"].str.strip() != ""
    )

    # Clean narratives
    df["narrative_clean"] = df["consumer_complaint_narrative"].apply(clean_text)

    # Filter dataset
    filtered = df.loc[
        df["product_std"].isin([p.title() for p in TARGET_PRODUCTS])
        & (df["narrative_clean"].str.strip() != ""),
        [
            "complaint_id",
            "product_std",
            "issue",
            "company",
            "state",
            "date_received",
            "narrative_clean",
        ],
    ].rename(columns={"product_std": "product"})

    # Save output
    os.makedirs(os.path.dirname(out_path), exist_ok=True)
    filtered.to_csv(out_path, index=False)
    print(f"Saved cleaned dataset: {out_path}")
    return filtered

# Run preprocessing
filtered_df = preprocess(RAW_PATH, OUT_PATH)

# Quick checka               
print(filtered_df.shape)
print(filtered_df.head())
```
